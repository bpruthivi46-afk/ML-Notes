```python
import numpy as np
```

```python
sl = pd.read_csv('data.csv')
sl.head()
age = sl['age'].to_list()
eng = sl['engineering'].to_list()
mar = sl['marketing'].to_list()
hr = sl['hr'].to_list()
```

```python
plt.bar(age,eng,label = 'Engineering')
plt.bar(age,mar,label = 'Marketing')
plt.bar(age,hr,label = 'HR')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.legend()
plt.show()
```

```python
x_index = np.arange(len(age))
width = 0.25
plt.bar(x_index - width,eng,width=width,label = 'Engineering',hatch = 'x')
plt.bar(x_index,mar,width=width,label = 'Marketing',alpha = 0.6)
plt.bar(x_index + width,hr,width=width,label = 'HR')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.title('Age vs Salary')
plt.legend()
plt.xticks(ticks=x_index,labels=age)
plt.show()
```

```python
df = pd.DataFrame({
    'name':     ['Ash', 'Misty', 'Brock', 'Gary'],
    'language': ['Python, Java', 'Python', 'Java, C++', 'Python, C++']
})
```

```python
from collections import Counter
```

```python
all_languages = df['language'].apply(lambda x: x.split(', '))
counts = Counter([lang for sublist in all_languages for lang in sublist])
print(counts)
```

```python
plt.style.use('ggplot')
y = list(counts.values())
x = list(counts.keys())
plt.barh(x,y,edgecolor='black',linewidth=1.5)
plt.xlabel('Count')
plt.title('Programming Language Popularity')
plt.yticks(rotation=45)
plt.show() 
```

```python
languages = ['Python', 'JavaScript', 'Java', 'C++', 'C#', 'TypeScript', 'Rust', 'Go', 'Kotlin', 'Swift']
popularity = [68500, 62300, 45800, 38200, 34700, 29400, 18600, 15300, 12800, 11200]
counts = {}
for i in range(len(languages)):
    counts.update({languages[i]: popularity[i]})    
print(counts)
```

```python
plt.style.use('ggplot')
explode = [0.1 if x[1] == max(counts.values()) else 0.05 for x in counts.items()]
plt.pie(list(counts.values()), labels=list(counts.keys()), autopct='%1.2f%%', startangle=0,explode = explode,wedgeprops ={'edgecolor': 'black','linewidth': 1})
plt.title('Programming Language Popularity')
plt.show()
```

```python
ages = [25, 30, 35, 40, 45, 50, 55, 60]
bins = [20, 30, 40, 50, 60]
plt.hist(ages, bins=bins, edgecolor='black', linewidth=1.5, color = '#f7e9d3')
plt.xlabel('Age')
plt.ylabel('frequency')
plt.title('Age Distribution')
plt.axvline(np.median(ages),color='black',linestyle='dashed',label='Median')
plt.legend()
plt.show()
```
