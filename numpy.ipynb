```python
c = np.random.randint(1,20,size=(15,17))
d = np.random.randint(1,20,size=(17,15))
print(c)
print(d)
```

Linear algebra


```python
e = np.matmul(c,d)
print(e)
np.linalg.det(e)
```

```python
print(np.linalg.eigvals(e))
print(np.linalg.diagonal(e))
print(np.linalg.inv(e))
```
